# Step 10 — Patient-Grouped Cross-Validation and Model Comparison

Logistic Regression and Random Forest were evaluated using
patient-grouped cross-validation. XGBoost was additionally evaluated
when available.

StratifiedGroupKFold was used to preserve approximate outcome balance
while ensuring that ICU stays belonging to the same patient were never
distributed between training and validation within a fold.

All missing-value imputation and scaling were performed within each
training fold to prevent information leakage.

Given the small MIMIC-IV Demo cohort and limited number of deterioration
events, model performance is considered exploratory. Fold-level
performance and variability are reported rather than relying on a
single train-test split.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    brier_score_loss,
)

DATA_PATH = "../results/model_features_2h.csv"
TARGET = "future_deterioration"
ID_COLUMNS = ["subject_id", "hadm_id", "stay_id"]
N_SPLITS = 5


df = pd.read_csv(DATA_PATH)

feature_columns = [
    column for column in df.columns
    if column not in ID_COLUMNS + [TARGET]
]
X = df[feature_columns].copy()
y = df[TARGET].astype(int).copy()
groups = df["subject_id"].copy()

print("=" * 70)
print("STEP 10 - GROUPED MODEL COMPARISON")
print("=" * 70)
print("ICU stays:", len(df))
print("Patients:", groups.nunique())
print("Predictors:", X.shape[1])
print("Positive outcomes:", int(y.sum()))
print("Negative outcomes:", int((y == 0).sum()))

all_missing_columns = [
    column for column in X.columns
    if X[column].isna().all()
]

if all_missing_columns:
    X = X.drop(columns=all_missing_columns)

feature_columns = X.columns.tolist()
print("Completely missing predictors:", all_missing_columns)
print("Predictors retained:", len(feature_columns))

cv = StratifiedGroupKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=42,
)

fold_structure = []
for fold, (train_idx, validation_idx) in enumerate(
    cv.split(X, y, groups),
    start=1,
):
    train_patients = set(groups.iloc[train_idx])
    validation_patients = set(groups.iloc[validation_idx])
    fold_structure.append({
        "fold": fold,
        "train_stays": len(train_idx),
        "validation_stays": len(validation_idx),
        "train_patients": len(train_patients),
        "validation_patients": len(validation_patients),
        "train_positive": int(y.iloc[train_idx].sum()),
        "validation_positive": int(y.iloc[validation_idx].sum()),
        "patient_overlap": len(train_patients & validation_patients),
    })

fold_structure = pd.DataFrame(fold_structure)
display(fold_structure)
assert (fold_structure["patient_overlap"] == 0).all()
print("All folds have zero patient overlap.")

logistic_model = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        (
            "model",
            LogisticRegression(
                class_weight="balanced",
                max_iter=1000,
                random_state=42,
            ),
        ),
    ]
)

random_forest_model = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        (
            "model",
            RandomForestClassifier(
                n_estimators=500,
                max_depth=4,
                min_samples_leaf=3,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1,
            ),
        ),
    ]
)

models = {
    "Logistic Regression": logistic_model,
    "Random Forest": random_forest_model,
}

try:
    from xgboost import XGBClassifier
except ImportError:
    XGBClassifier = None

if XGBClassifier is not None:
    models["XGBoost"] = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median")),
            (
                "model",
                XGBClassifier(
                    n_estimators=150,
                    max_depth=2,
                    learning_rate=0.03,
                    subsample=0.8,
                    colsample_bytree=0.8,
                    objective="binary:logistic",
                    eval_metric="logloss",
                    random_state=42,
                    n_jobs=-1,
                ),
            ),
        ]
    )
    print("XGBoost is available.")
else:
    print("XGBoost is not installed.")

print("Models being evaluated:")
for model_name in models:
    print("-", model_name)


def evaluate_grouped_cv(model, model_name, X, y, groups, cv):
    fold_results = []
    prediction_results = []

    for fold, (train_idx, validation_idx) in enumerate(
        cv.split(X, y, groups),
        start=1,
    ):
        fold_model = clone(model)
        X_train = X.iloc[train_idx]
        X_validation = X.iloc[validation_idx]
        y_train = y.iloc[train_idx]
        y_validation = y.iloc[validation_idx]

        fold_model.fit(X_train, y_train)
        probabilities = fold_model.predict_proba(X_validation)[:, 1]
        predictions = (probabilities >= 0.5).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_validation,
            predictions,
            labels=[0, 1],
        ).ravel()
        train_patients = set(groups.iloc[train_idx])
        validation_patients = set(groups.iloc[validation_idx])

        fold_results.append({
            "model": model_name,
            "fold": fold,
            "validation_stays": len(validation_idx),
            "validation_positives": int(y_validation.sum()),
            "patient_overlap": len(train_patients & validation_patients),
            "accuracy": accuracy_score(y_validation, predictions),
            "balanced_accuracy": balanced_accuracy_score(
                y_validation, predictions
            ),
            "precision": precision_score(
                y_validation, predictions, zero_division=0
            ),
            "sensitivity": recall_score(
                y_validation, predictions, zero_division=0
            ),
            "specificity": tn / (tn + fp) if (tn + fp) else 0.0,
            "f1": f1_score(y_validation, predictions, zero_division=0),
            "auroc": (
                roc_auc_score(y_validation, probabilities)
                if y_validation.nunique() > 1
                else np.nan
            ),
            "pr_auc": average_precision_score(
                y_validation, probabilities
            ),
            "brier": brier_score_loss(y_validation, probabilities),
        })

        prediction_results.append(pd.DataFrame({
            "model": model_name,
            "fold": fold,
            "index": X_validation.index,
            "actual": y_validation.to_numpy(),
            "predicted_probability": probabilities,
            "predicted_class": predictions,
        }))

    return (
        pd.DataFrame(fold_results),
        pd.concat(prediction_results, ignore_index=True),
    )


all_fold_results = []
all_predictions = []

for model_name, model in models.items():
    print(f"Evaluating {model_name}...")
    fold_result, predictions = evaluate_grouped_cv(
        model=model,
        model_name=model_name,
        X=X,
        y=y,
        groups=groups,
        cv=cv,
    )
    all_fold_results.append(fold_result)
    all_predictions.append(predictions)

cv_results = pd.concat(all_fold_results, ignore_index=True)
cv_predictions = pd.concat(all_predictions, ignore_index=True)

print("\nCross-validation complete.")
display(cv_results)

metric_columns = [
    "accuracy",
    "balanced_accuracy",
    "precision",
    "sensitivity",
    "specificity",
    "f1",
    "auroc",
    "pr_auc",
    "brier",
]

model_summary = cv_results.groupby("model")[metric_columns].agg(["mean", "std"])
display(model_summary.round(3))

model_comparison = (
    cv_results.groupby("model")[metric_columns]
    .agg(["mean", "std"])
)
model_comparison.columns = [
    f"{metric}_{statistic}"
    for metric, statistic in model_comparison.columns
]
model_comparison = model_comparison.reset_index()
display(model_comparison.round(3))

pooled_results = []
for model_name, model_predictions in cv_predictions.groupby("model"):
    actual = model_predictions["actual"]
    probability = model_predictions["predicted_probability"]
    predicted = model_predictions["predicted_class"]
    tn, fp, fn, tp = confusion_matrix(
        actual,
        predicted,
        labels=[0, 1],
    ).ravel()
    pooled_results.append({
        "model": model_name,
        "auroc": roc_auc_score(actual, probability),
        "pr_auc": average_precision_score(actual, probability),
        "accuracy": accuracy_score(actual, predicted),
        "balanced_accuracy": balanced_accuracy_score(actual, predicted),
        "sensitivity": recall_score(actual, predicted, zero_division=0),
        "specificity": tn / (tn + fp) if (tn + fp) else 0.0,
        "precision": precision_score(actual, predicted, zero_division=0),
        "f1": f1_score(actual, predicted, zero_division=0),
        "brier": brier_score_loss(actual, probability),
    })

pooled_results = pd.DataFrame(pooled_results)
display(pooled_results.round(3))

for model_name, subset in cv_results.groupby("model"):
    plt.figure(figsize=(6, 4))
    plt.bar(subset["fold"].astype(str), subset["auroc"])
    plt.axhline(0.5, linestyle="--")
    plt.ylim(0, 1)
    plt.xlabel("Fold")
    plt.ylabel("AUROC")
    plt.title(f"{model_name} - AUROC Across Folds")
    plt.tight_layout()
    plt.show()

plot_data = model_comparison[
    ["model", "auroc_mean", "pr_auc_mean", "balanced_accuracy_mean"]
].set_index("model")
plot_data.plot(kind="bar", figsize=(9, 5))
plt.ylabel("Mean cross-validation score")
plt.title("Grouped Cross-Validation Model Comparison")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("\n" + "=" * 70)
print("STEP 10 - GROUPED MODEL COMPARISON COMPLETE")
print("=" * 70)
print("ICU stays:", len(df))
print("Patients:", groups.nunique())
print("Positive outcomes:", int(y.sum()))
print("Cross-validation folds:", N_SPLITS)
print("Models evaluated:", ", ".join(models.keys()))
print("Maximum patient overlap:", int(cv_results["patient_overlap"].max()))
print("\nMEAN +/- SD ACROSS FOLDS")
print("-" * 70)

for model_name, subset in cv_results.groupby("model"):
    print(f"\n{model_name}")
    for metric, label in [
        ("auroc", "AUROC"),
        ("pr_auc", "PR-AUC"),
        ("balanced_accuracy", "Balanced accuracy"),
        ("sensitivity", "Sensitivity"),
        ("specificity", "Specificity"),
        ("brier", "Brier score"),
    ]:
        print(
            f"{label}: {subset[metric].mean():.3f} +/- "
            f"{subset[metric].std():.3f}"
        )

print("\nNo patient overlap across folds.")
print("Preprocessing performed within training folds.")
print("Results are exploratory because of the small Demo cohort.")
```
