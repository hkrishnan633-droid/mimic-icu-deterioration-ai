# Step 8 — Exploratory Data Analysis

Exploratory data analysis was performed on the feature dataset derived
from the first 2 hours following ICU admission.

The objectives were to:

1. verify cohort and outcome distributions;
2. quantify missing data;
3. examine predictor distributions;
4. identify potential outliers and physiologically implausible values;
5. compare predictor distributions between deterioration and
   non-deterioration groups;
6. assess correlation and redundancy among candidate predictors; and
7. identify potential data-quality issues before model development.

No imputation or model fitting was performed during this stage.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_PATH = "../results/model_features_2h.csv"

df = pd.read_csv(DATA_PATH)

print("=" * 70)
print("STEP 8 — EXPLORATORY DATA ANALYSIS")
print("=" * 70)

print("Dataset shape:", df.shape)
print("Unique ICU stays:", df["stay_id"].nunique())
print("Unique patients:", df["subject_id"].nunique())
print("Duplicate stay IDs:", df["stay_id"].duplicated().sum())

print("\nOutcome distribution:")
print(df["future_deterioration"].value_counts())

print("\nOutcome percentage:")
print(
    df["future_deterioration"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

ID_COLUMNS = [
    "subject_id",
    "hadm_id",
    "stay_id"
]

TARGET = "future_deterioration"

feature_columns = [
    col for col in df.columns
    if col not in ID_COLUMNS + [TARGET]
]

print("Number of predictor columns:", len(feature_columns))

outcome_counts = (
    df[TARGET]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(6, 4))

plt.bar(
    ["No deterioration", "Deterioration"],
    [
        outcome_counts.get(0, 0),
        outcome_counts.get(1, 0)
    ]
)

plt.ylabel("Number of ICU stays")
plt.title("Clinical Deterioration Outcome Distribution")
plt.tight_layout()
plt.show()

n_negative = (df[TARGET] == 0).sum()
n_positive = (df[TARGET] == 1).sum()

print("Negative cases:", n_negative)
print("Positive cases:", n_positive)

print(
    "Positive event rate:",
    round(n_positive / len(df) * 100, 2),
    "%"
)

print(
    "Negative-to-positive ratio:",
    round(n_negative / n_positive, 2),
    ": 1"
)

missing_summary = pd.DataFrame({
    "missing_count":
        df[feature_columns].isna().sum(),

    "missing_percent":
        df[feature_columns]
        .isna()
        .mean()
        .mul(100)
}).sort_values(
    "missing_percent",
    ascending=False
)

missing_summary["missing_percent"] = (
    missing_summary["missing_percent"]
    .round(2)
)

display(missing_summary)

missing_summary.to_csv(
    "../results/eda_missingness.csv"
)

top_missing = (
    missing_summary
    .head(20)
    .sort_values("missing_percent")
)

plt.figure(figsize=(9, 7))

plt.barh(
    top_missing.index,
    top_missing["missing_percent"]
)

plt.xlabel("Missing (%)")
plt.ylabel("Feature")
plt.title("Top 20 Features by Missingness")
plt.tight_layout()
plt.show()

df["number_missing_features"] = (
    df[feature_columns]
    .isna()
    .sum(axis=1)
)

df["percent_missing_features"] = (
    df["number_missing_features"]
    / len(feature_columns)
    * 100
)

display(
    df[
        [
            "subject_id",
            "stay_id",
            TARGET,
            "number_missing_features",
            "percent_missing_features"
        ]
    ]
    .sort_values(
        "percent_missing_features",
        ascending=False
    )
    .head(20)
)

print(
    df["percent_missing_features"]
    .describe()
)

missing_indicator_columns = [
    col for col in feature_columns
    if col.endswith("_missing")
]

clinical_feature_columns = [
    col for col in feature_columns
    if col not in missing_indicator_columns
]

print(
    "Clinical feature columns:",
    len(clinical_feature_columns)
)

print(
    "Missingness indicator columns:",
    len(missing_indicator_columns)
)

descriptive_stats = (
    df[clinical_feature_columns]
    .describe()
    .T
)

descriptive_stats["missing_percent"] = (
    df[clinical_feature_columns]
    .isna()
    .mean()
    .mul(100)
)

display(descriptive_stats)

descriptive_stats.to_csv(
    "../results/eda_descriptive_statistics.csv"
)

core_vital_terms = [
    "heart_rate",
    "respiratory_rate",
    "systolic_bp",
    "diastolic_bp"
]

core_vital_features = [
    col for col in clinical_feature_columns
    if any(
        term in col
        for term in core_vital_terms
    )
]

print(
    "Core vital features:",
    len(core_vital_features)
)

display(
    df[core_vital_features]
    .describe()
    .T
)

quality_flags = pd.DataFrame(index=df.index)

if "heart_rate_min" in df.columns:
    quality_flags["hr_zero_or_less"] = df["heart_rate_min"] <= 0

if "respiratory_rate_min" in df.columns:
    quality_flags["rr_zero_or_less"] = df["respiratory_rate_min"] <= 0

if "systolic_bp_min" in df.columns:
    quality_flags["sbp_zero_or_less"] = df["systolic_bp_min"] <= 0

if "diastolic_bp_min" in df.columns:
    quality_flags["dbp_zero_or_less"] = df["diastolic_bp_min"] <= 0

print("Quality-control flags:")
for col in quality_flags.columns:
    print(col, ":", int(quality_flags[col].sum()))

group_summary = (
    df.groupby(TARGET)[
        clinical_feature_columns
    ]
    .median()
    .T
)

group_summary.columns = [
    "median_no_deterioration",
    "median_deterioration"
]

display(group_summary)

group_summary.to_csv(
    "../results/eda_group_medians.csv"
)

preferred_features = [
    "heart_rate_mean",
    "respiratory_rate_mean",
    "systolic_bp_mean",
    "diastolic_bp_mean",
    "creatinine_last",
    "lactate_last",
    "wbc_last",
    "hemoglobin_last",
    "platelets_last",
    "sodium_last",
    "potassium_last",
    "bicarbonate_last",
    "glucose_last"
]

plot_features = [
    col for col in preferred_features
    if col in df.columns
]

print("Features available for plotting:")
print(plot_features)

for feature in plot_features:
    group0 = df.loc[
        df[TARGET] == 0,
        feature
    ].dropna()

    group1 = df.loc[
        df[TARGET] == 1,
        feature
    ].dropna()

    plt.figure(figsize=(6, 4))
    plt.boxplot([group0, group1])
    plt.xticks([1, 2], ["No deterioration", "Deterioration"])
    plt.ylabel(feature)
    plt.title(f"{feature}: Outcome Comparison")
    plt.tight_layout()
    plt.show()

correlation_matrix = (
    df[clinical_feature_columns]
    .corr()
)

display(
    correlation_matrix.round(2)
)

high_corr_pairs = []

columns = correlation_matrix.columns

for i in range(len(columns)):
    for j in range(i + 1, len(columns)):

        corr_value = correlation_matrix.iloc[i, j]

        if (
            pd.notna(corr_value)
            and abs(corr_value) >= 0.90
        ):

            high_corr_pairs.append({
                "feature_1": columns[i],
                "feature_2": columns[j],
                "correlation": round(
                    corr_value,
                    3
                )
            })

high_corr_df = pd.DataFrame(high_corr_pairs)

if len(high_corr_df) > 0:
    high_corr_df = (
        high_corr_df
        .sort_values(
            "correlation",
            key=abs,
            ascending=False
        )
    )
    display(high_corr_df)
else:
    print("No feature pairs with |correlation| >= 0.90")

high_corr_df.to_csv(
    "../results/eda_high_correlations.csv",
    index=False
)

unique_counts = (
    df[feature_columns]
    .nunique(dropna=True)
    .sort_values()
)

constant_features = (
    unique_counts[
        unique_counts <= 1
    ]
)

print("Constant features:")
print(constant_features)

missing_by_outcome = []

for feature in clinical_feature_columns:

    for outcome_value in [0, 1]:

        subset = df[
            df[TARGET] == outcome_value
        ]

        missing_pct = (
            subset[feature]
            .isna()
            .mean()
            * 100
        )

        missing_by_outcome.append({
            "feature": feature,
            "outcome": outcome_value,
            "missing_percent": round(
                missing_pct,
                2
            )
        })

missing_by_outcome = pd.DataFrame(
    missing_by_outcome
)

missing_pivot = (
    missing_by_outcome
    .pivot(
        index="feature",
        columns="outcome",
        values="missing_percent"
    )
)

missing_pivot.columns = [
    "missing_no_deterioration",
    "missing_deterioration"
]

missing_pivot[
    "absolute_difference"
] = (
    missing_pivot[
        "missing_deterioration"
    ]
    -
    missing_pivot[
        "missing_no_deterioration"
    ]
).abs()

missing_pivot = (
    missing_pivot
    .sort_values(
        "absolute_difference",
        ascending=False
    )
)

display(
    missing_pivot.head(20)
)

available_feature_count = (
    df[clinical_feature_columns]
    .notna()
    .sum(axis=1)
)

print(
    "Available clinical features per ICU stay:"
)

print(
    available_feature_count.describe()
)

plt.figure(figsize=(7, 4))

plt.hist(
    available_feature_count,
    bins=15
)

plt.xlabel(
    "Number of available clinical features"
)
plt.ylabel(
    "Number of ICU stays"
)
plt.title(
    "Feature Availability per ICU Stay"
)

plt.tight_layout()
plt.show()

eda_summary = pd.DataFrame({
    "metric": [
        "ICU stays",
        "Unique patients",
        "Positive outcomes",
        "Negative outcomes",
        "Event rate (%)",
        "Candidate predictors",
        "Clinical feature columns",
        "Missingness indicators",
        "Constant features",
        "Highly correlated pairs"
    ],

    "value": [
        len(df),
        df["subject_id"].nunique(),
        int(df[TARGET].sum()),
        int((df[TARGET] == 0).sum()),
        round(df[TARGET].mean() * 100, 2),
        len(feature_columns),
        len(clinical_feature_columns),
        len(missing_indicator_columns),
        len(constant_features),
        len(high_corr_df)
    ]
})

display(eda_summary)

eda_summary.to_csv(
    "../results/eda_summary.csv",
    index=False
)

if len(quality_flags.columns) > 0:
    quality_output = pd.concat(
        [
            df[
                [
                    "subject_id",
                    "stay_id",
                    TARGET
                ]
            ],
            quality_flags
        ],
        axis=1
    )

    quality_output.to_csv(
        "../results/eda_quality_flags.csv",
        index=False
    )

print("=" * 70)
print("STEP 8 — EDA COMPLETE")
print("=" * 70)

print("ICU stays:", len(df))
print(
    "Unique patients:",
    df["subject_id"].nunique()
)

print(
    "Positive outcomes:",
    int(df[TARGET].sum())
)

print(
    "Negative outcomes:",
    int((df[TARGET] == 0).sum())
)

print(
    "Event rate:",
    round(
        df[TARGET].mean() * 100,
        2
    ),
    "%"
)

print(
    "Candidate predictors:",
    len(feature_columns)
)

print(
    "Clinical features:",
    len(clinical_feature_columns)
)

print(
    "Missingness indicators:",
    len(missing_indicator_columns)
)

print(
    "Constant features:",
    len(constant_features)
)

print(
    "Highly correlated feature pairs:",
    len(high_corr_df)
)

print("\nNo imputation performed.")
print("No model fitted.")
print("Original Step 7 dataset preserved.")

print("\nSaved EDA outputs to ../results/")

missing_summary.head(10)
high_corr_df.head(10)

display(missing_summary.head(10))
display(high_corr_df.head(10))

print("=" * 70)
print("STEP 8 — EDA COMPLETE")
print("=" * 70)

print("ICU stays:", len(df))
print("Unique patients:", df["subject_id"].nunique())
print("Positive outcomes:", int(df[TARGET].sum()))
print("Negative outcomes:", int((df[TARGET] == 0).sum()))
print("Event rate:", round(df[TARGET].mean() * 100, 2), "%")
print("Candidate predictors:", len(feature_columns))
print("Clinical features:", len(clinical_feature_columns))
print("Missingness indicators:", len(missing_indicator_columns))
print("Constant features:", len(constant_features))
print("Highly correlated feature pairs:", len(high_corr_df))
```
